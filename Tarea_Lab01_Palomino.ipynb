## 4.3. Clasificación de sistemas con Inteligencia Artificial

Para esta actividad se seleccionaron dos sistemas de Inteligencia Artificial:

1. Sistema de detección de fraude bancario en tiempo real.
2. Sistema de autenticación biométrica.

### Sistema 1: Detección de fraude bancario en tiempo real

| Aspecto | Descripción |
|---|---|
| **Entrada** | Monto de la transacción, hora, ubicación, historial de operaciones y datos de la transacción. |
| **Salida** | Clasificación de la transacción como legítima o sospechosa/fraudulenta. |
| **Tipo de problema** | Clasificación. |
| **Área de IA** | Reconocimiento de patrones y analítica predictiva. |

**Razón:** El sistema analiza características de cada transacción y busca patrones asociados con operaciones fraudulentas. Por ello, el problema principal es de clasificación, ya que debe determinar a qué categoría pertenece una transacción.

### Sistema 2: Autenticación biométrica

| Aspecto | Descripción |
|---|---|
| **Entrada** | Imagen o características biométricas de una persona, como el rostro o la huella digital. |
| **Salida** | Identificación o decisión sobre si la persona coincide con el registro almacenado. |
| **Tipo de problema** | Clasificación. |
| **Área de IA** | Reconocimiento de patrones. |

**Razón:** El sistema compara las características biométricas recibidas con patrones previamente registrados para determinar si existe una coincidencia. Por ello, se considera principalmente un problema de clasificación y reconocimiento de patrones.

## 4.4. Análisis PEAS de un sistema de Inteligencia Artificial

### Sistema seleccionado: Detección de fraude bancario en tiempo real

El objetivo del sistema es analizar las transacciones bancarias y detectar
operaciones que puedan ser fraudulentas, permitiendo tomar una decisión
rápida para proteger al usuario y reducir pérdidas económicas.

### Análisis PEAS

| Elemento | Descripción |
|---|---|
| **P - Percepción** | Monto de la transacción, hora, ubicación, dispositivo utilizado, historial de operaciones y frecuencia de transacciones. |
| **A - Acciones** | Aprobar la transacción, bloquearla temporalmente o solicitar una verificación adicional al usuario. |
| **E - Entorno** | Sistema bancario, usuarios, cuentas bancarias, comercios y transacciones realizadas en tiempo real. |
| **S - Medida de desempeño** | Detectar correctamente las operaciones fraudulentas, reducir falsos positivos, responder rápidamente y disminuir pérdidas económicas. |

### Tipo de agente

El sistema puede considerarse principalmente un **agente de aprendizaje**, 
porque puede utilizar datos históricos de transacciones para aprender patrones
relacionados con operaciones legítimas y fraudulentas.

### ¿Por qué es racional?

El agente es racional porque utiliza la información disponible sobre cada
transacción para seleccionar una acción que ayude a reducir el riesgo de fraude.
Su decisión busca proteger los recursos del usuario y mantener la seguridad
del sistema bancario.

### ¿Es IA estrecha o IA general?

Es un sistema de **IA estrecha**, porque está diseñado para realizar una tarea
específica: detectar posibles operaciones fraudulentas. No posee capacidades
generales para resolver cualquier tipo de problema intelectual como una persona.
