```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    PEAS:
    - Percepción (P): temperatura actual medida por el sensor.
    - Acciones (A): encender/apagar calefacción, encender/apagar aire acondicionado.
    - Entorno (E): casa, habitación u oficina.
    - Medida de desempeño (S): eficiencia al regular la temperatura,
      tiempo necesario para regularla y ahorro energético.
    """

    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        return "Apagar la calefacción y encender el aire acondicionado"
    elif diferencia < -margen:
        return "Encender la calefacción y apagar el aire acondicionado"
    else:
        return "Apagar todo el sistema"
```

## 5.6. Simulación del agente termostato

Se probará el agente con diferentes valores de temperatura para observar
cómo responde ante condiciones de frío, temperatura adecuada y calor.

```python
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepción: {t} °C -> Acción del agente: {accion}")
```

### Pruebas adicionales

Se agregan dos valores de temperatura propios para comprobar el comportamiento
del agente en situaciones diferentes. También se incluye un caso límite,
considerando el margen de 1 °C respecto a la temperatura objetivo de 22 °C.

```python
pruebas_adicionales = [16.0, 23.0]

for t in pruebas_adicionales:
    accion = agente_termostato(t)
    print(f"Percepción: {t} °C -> Acción del agente: {accion}")
```

### Reflexión

El agente termostato puede considerarse racional porque selecciona una acción
de acuerdo con la temperatura actual y la temperatura objetivo. Si hace
demasiado frío, activa la calefacción; si hace demasiado calor, activa el aire
acondicionado; y si la temperatura se encuentra dentro del margen permitido,
mantiene ambos sistemas apagados.

Sin embargo, el agente podría dejar de ser racional si el sensor de temperatura
entrega información incorrecta o si el margen establecido no representa
adecuadamente las necesidades del ambiente. Por ejemplo, una medición
incorrecta podría hacer que el agente tome una decisión equivocada.
